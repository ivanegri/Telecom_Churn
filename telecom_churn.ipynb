## Preditive modeling:



* Wich Machine Learning algorithm (Logistic Regression, Random Forest, XGBoost) better perform performe?


```python
X = dfmodel.drop(columns={'churn'})
```

```python
X
```

```python
y=dfmodel['churn'].reset_index()
```

```python
y.reset_index(inplace=True)
```

```python
y=y.drop(columns={'level_0','index'})
```

```python
from sklearn.model_selection import train_test_split
```

### 25% of data for test

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.25, 
    random_state=42,
    stratify=y
   )
print('Treino',round(y_train.mean().mul(100),2))
print('Teste',round(y_test.mean().mul(100),2))
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
print("\nShape dos dados:")
print(f"X_train: {X_train.shape} | y_train: {y_train.shape}")
print(f"X_test:  {X_test.shape}  | y_test:  {y_test.shape}")

print("\nExemplo de X_train:")
print(X_train.head(3))
```

### Precision Rate

```python
from xgboost import XGBClassifier
from sklearn.metrics import classification_report

# Treinar modelo
model = XGBClassifier(
    scale_pos_weight=(len(y_train.churn) - sum(y_train.churn)) / sum(y_train.churn),  # Balanceamento
    eval_metric='aucpr',
    random_state=42
)
model.fit(X_train_scaled, y_train)

# Avaliar
y_pred = model.predict(X_test_scaled)
print(classification_report(y_test, y_pred))
```

### Bad precision Model
